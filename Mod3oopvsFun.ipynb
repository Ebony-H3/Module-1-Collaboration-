```python


def sort012(arr):
    c0 = 0
    c1 = 0
    c2 = 0

    # Count 0s, 1s and 2s
    for num in arr:
        if num == 0:
            c0 += 1
        elif num == 1:
            c1 += 1
        else:
            c2 += 1

    idx = 0

    # Place all the 0s
    for i in range(c0):
        arr[idx] = 0
        idx += 1

    # Place all the 1s
    for i in range(c1):
        arr[idx] = 1
        idx += 1

    # Place all the 2s
    for i in range(c2):
        arr[idx] = 2
        idx += 1

test_array = [2,0,1,2,0,1]
sort012(test_array)
print("Sorted Array:", test_array)
```

```python
class Solution:
    def firstSearch(self, arr, k):
        l = 0
        r = len(arr) - 1
        pos = -1
        
        while l <= r:
            mid = (l + r) // 2
            
            if arr[mid] == k:
                pos = mid
                r = mid - 1  #searches left for the first occurrence
            elif arr[mid] > k:
                r = mid - 1
            else:
                l = mid + 1
                
        return pos


#Initialize the class
sol = Solution()

#Set up test parameters (Array has duplicate 2s)
test_array = [1, 2, 2, 2, 3, 4, 5]
target_value = 2

#Call the function
result = sol.firstSearch(test_array, target_value)

#Print the result
print("The first occurrence of", target_value, "is at index:", result)

```
